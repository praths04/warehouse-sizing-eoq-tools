```python
import math

# Function to calculate total cost
def calculate_total_cost(demand, ordering_cost, carrying_cost_percentage, quantity_slabs, unit_cost, discount_percentages):
    try:
        # Calculate unit costs for each slab
        unit_costs = [unit_cost * (1 - discount) for discount in discount_percentages]
        
        # Calculate carrying costs for each slab
        carrying_costs = [unit_cost * carrying_cost_percentage for unit_cost in unit_costs]
        
        # Calculate EOQ for each slab and optimize if necessary
        eoqs = []
        optimized_eoqs = []
        for i, quantity_slab in enumerate(quantity_slabs):
            eoq = int(math.sqrt((2 * demand * ordering_cost) / carrying_costs[i]))
            if eoq < quantity_slab:
                optimized_eoq = optimize_eoq(demand, ordering_cost, unit_costs[i], carrying_costs[i], quantity_slab)
            else:
                optimized_eoq = eoq
            eoqs.append(eoq)
            optimized_eoqs.append(optimized_eoq)
        
        # Calculate total cost for each slab
        total_costs = []
        for i in range(len(quantity_slabs)):
            total_cost = (unit_costs[i] * demand) + (optimized_eoqs[i] / 2 * carrying_costs[i]) + (demand / optimized_eoqs[i] * ordering_cost)
            total_costs.append(total_cost)
        
        # Find the minimum total cost
        min_total_cost = min(total_costs)
        min_total_cost_index = total_costs.index(min_total_cost)
        
        # Display the results
        print(f"Quantity Slab: {quantity_slabs[min_total_cost_index]}, Optimized EOQ: {optimized_eoqs[min_total_cost_index]}, Minimum Total Cost: {min_total_cost:.2f}")
    
    except ValueError:
        print("Invalid input. Please enter valid numbers.")

# Function to optimize EOQ
def optimize_eoq(demand, ordering_cost, unit_cost, carrying_cost, quantity_slab):
    # Optimize EOQ to minimize total cost
    min_total_cost = float('inf')
    optimized_eoq = 0
    for eoq in range(1, demand + 1):
        if eoq >= quantity_slab:
            total_cost = (unit_cost * demand) + (eoq / 2 * carrying_cost) + (demand / eoq * ordering_cost)
        else:
            if optimized_eoq != 0:  # Ensure optimized_eoq is not zero before division
                total_cost = (unit_cost * demand) + (optimized_eoq / 2 * carrying_cost) + (demand / optimized_eoq * ordering_cost)
            else:
                total_cost = float('inf')  # Set total_cost to infinity to avoid division by zero
        if total_cost < min_total_cost:
            min_total_cost = total_cost
            optimized_eoq = eoq
    return optimized_eoq


# Input values from the user
demand = int(input("Enter demand: "))
ordering_cost = float(input("Enter ordering cost: "))
carrying_cost_percentage = float(input("Enter carrying cost percentage (e.g., 20% as 0.2): "))
num_slabs = int(input("Enter the number of quantity slabs: "))

quantity_slabs = []
discount_percentages = []

print("Enter quantity slabs and corresponding discount percentages:")
for _ in range(num_slabs):
    slab_input = input("Enter quantity slab range (e.g., '500-1000'): ")
    slab_values = [int(value.strip()) for value in slab_input.split()]  # Extract numeric values
    if len(slab_values) == 1:
        # If only one value is entered, treat it as a single value slab
        quantity_slabs.append(slab_values[0])
    elif len(slab_values) == 2:
        # If two values are entered, calculate the average as the slab value
        slab_average = sum(slab_values) / 2
        quantity_slabs.append(slab_average)
    else:
        print("Invalid input. Please enter a single value or a range.")
        continue
    
    discount_percent = float(input("Enter discount percentage for this slab (e.g., 20% as 0.2): "))
    discount_percentages.append(discount_percent)

unit_cost = float(input("Enter unit cost: "))

# Calculate total cost
calculate_total_cost(demand, ordering_cost, carrying_cost_percentage, quantity_slabs, unit_cost, discount_percentages)
```
